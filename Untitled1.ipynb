# Calculadora de Comissão do Vendedor
# Objetivo: Calcular valor que sera recebido de comissao

```python
# Interação com os Usuario
vendedor = input('Nome do vendedor:')
meta = float(input("Digite o valor da meta (R$): "))
vendas = float(input('Digite o valor das vendas R$:'))
percentual_comissao = float(input('Digite o percentual da comissao:'))
```

```python
# Calculo
comissao = vendas * percentual_comissao / 100
atingimento_meta = (vendas / meta) * 100
```

```python
# Exibição dos Resultados
print("\n=== RELATÓRIO DE COMISSÃO ===")
print(f"Vendedor: {vendedor}")
print(f"Valor das Vendas: R$ {vendas:.2f}")
print(f"Valor da Comissão: R$ {comissao:.2f}")
print(f"Total Vendido: R$ {vendas:,.2f}")
```

```python
# Mensagem de desempenho
if atingimento_meta >= 100:
    print(" Excelente! Meta batida!")
elif 80 <= atingimento_meta < 100:
    print(" Quase lá! Continue focado.")
else:
    print(" Abaixo da meta. Reveja suas estratégias.")
```
